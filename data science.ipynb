```python
# add an item in the list
Name=["Kiran","Anil","Lalith","Charig"]
Name.append("Deepika")
print(Name)
```

```python
#insert in particular index
student_name=["Arav","Kapil","Sanjeev"]
student_name.insert(1,"Hariharan")
print(student_name)
```

```python
# extend function
sec_a=["Chirag","Jhansi","Devi"]
sec_b=["Kiran","Anil","chari"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item from a list
sec_a_student=["chirag","anil","amit","devi"]
sec_a_student.remove("devi")
print(sec_a_student)
del sec_a_student[1]
print(sec_a_student)
sec_a_student.pop()
print(sec_a_student)
```

```python
# list slicing
subject=["python","statistics","sql","java","os","machine learning"]
print(subject[1:4])
print(subject[:3])
print(subject[4:])
print(subject[::2])
# length of the list
print("length of the list",len(subject))
```

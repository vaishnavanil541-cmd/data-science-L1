# Tuple
## ordered collection of items
## symbol ()
## Tuple Immitable
## supports index

```python
# for example
```

```python
# create a tuple
students=(101,"Chirag","cse",8.9)
print(students)

# access by index
print("Student Department:",students[2])
# access by negative index
print("Accessing by negative index:",students[-2])
```

```python
subjects=("DSV","Mechine Learning","OS","Cloud","SQL","Python")
print(subjects)
# find the lenght of tuple
print("Lenght of tuple:",len(subjects))
```

```python
# tuple is mutable or immutable
print(students)
students[1]="Manish"
print(students)
```

```python
# Nested tuple
student_info=(1005,"Rahul",(89,98,78,75))
print(student_info)
print("Nested tuple displaying 3rd subject mark:",student_info[2][2])
```

```python
Position=student_info.index("Rahul")
print("Data Rahul Presention the index",Position)
```

# SET
## unordered collection of items
## mutble
## No duplication
## It will give The Unique values
## Union, Intersection,Difference,Symmetric Difference,Super set

```python
# create a set
subject_info={"Date Science","Mechine Learning","OS","Cloud","SQL","Python"}
print(subject_info)
```

```python
studentsname={"Amit","Sanjiv","Kalai","Amit","Kiran","Neha","Sanjiv"}
print(studentsname)
```

```python
print(studentsname[2])
```

```python
# adding an element in the set
studentsname.add("Priya")
print(studentsname)
# update the set
studentsname.update(["Anil","Nandhani","Balaji"])
print(studentsname)
```

```python
# remove an item in the set
# studentsname.remove("Amit")
# print(studentsname)
# students
studentsname.discard("Priya")
print(studentsname)
studentsname.discard("Amit")
print(studentsname)
```

```python
# clear an delete
number={89,78,68,97,69}
number.clear()
print(number)
```

```python
# find the element in the class
print(studentsname)
 # find the element
print("Anil" in studentsname)
```

```python
# Set Operation
# Union
# Intersection
# Difference

python_students={"Rahul","Priya","Hari","Ganesh"}
java_students={"Rahul","Hari","Kiran","Ravi","Giri"}
res_union=python_students.union(java_students)
print(res_union)
```

```python
res_diff=python_students.difference(java_students)
print(res_diff)
```

```python
res_intersection=python_students.intersection(java_students)
print(res_intersection)
```

### Q50. Create a grouped summary by semester containing average attendance, average study hours, average predicted score, and number of students.

```python
semester_summary = df.groupby("semester").agg(
    average_attendance=("attendance_percent", "mean"),
    average_study_hours=("study_hours_per_day", "mean"),
    average_predicted_score=("predicted_score", "mean"),
    number_of_students=("student_id", "count")
).reset_index()

semester_summary
```
